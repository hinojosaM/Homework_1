```
import pandas as pd
import numpy as np
```

```
# Download the dataset using wget
!wget https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv

# Load the dataset into a DataFrame
df = pd.read_csv('car_fuel_efficiency_2026.csv')

# Inspect the first few rows
df.head()

```

```
# Q2: Count the total number of records
num_records = len(df)
print(f"Total number of records: {num_records}")
```

```
# Identify the fuel type column
fuel_column = 'fuel_type' if 'fuel_type' in df.columns else 'fuel'

# Q3: Count distinct fuel types
num_fuel_types = df[fuel_column].nunique()
print(f"Number of fuel types: {num_fuel_types}")
print("Unique fuel types:", df[fuel_column].unique())
```

```
# Check count of missing values per column
missing_per_column = df.isnull().sum()
print("Missing values per column:\n", missing_per_column)

# Q4: Count how many columns contain at least one missing value
cols_with_missing = (missing_per_column > 0).sum()
print(f"\nNumber of columns with missing values: {cols_with_missing}")
```

```
# Find the exact efficiency column name
efficiency_col = [col for col in df.columns if 'efficiency' in col or 'mpg' in col][0]

# Q5: Filter cars from Asia and find the maximum efficiency
max_asia_efficiency = df[df['origin'] == 'Asia'][efficiency_col].max()

print(f"Using column: {efficiency_col}")
print(f"Maximum fuel efficiency for cars from Asia: {max_asia_efficiency}")
```

```
# 1. Calculate the initial median
initial_median = df['horsepower'].median()
print(f"Initial median of horsepower: {initial_median}")

# 2. Calculate the most frequent value (mode)
horsepower_mode = df['horsepower'].mode()[0]
print(f"Most frequent value (mode): {horsepower_mode}")

# 3. Fill missing values with the mode on a copy of the dataframe
df_imputed = df.copy()
df_imputed['horsepower'] = df_imputed['horsepower'].fillna(horsepower_mode)

# 4. Calculate the median after imputation
updated_median = df_imputed['horsepower'].median()
print(f"Updated median of horsepower: {updated_median}")

# 5. Check if the median changed
if updated_median > initial_median:
    print("Answer: Yes, it increased")
elif updated_median < initial_median:
    print("Answer: Yes, it decreased")
else:
    print("Answer: No")
```

```
# 1 & 2. Filter cars from Asia and select 'vehicle_weight' and 'model_year'
asia_cars = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']]

# 3. Select the first 7 values
asia_subset = asia_cars.iloc[:7]

# 4. Convert to NumPy array X
X = asia_subset.to_numpy()
print("Matrix X shape:", X.shape)
print("Matrix X:\n", X)

# 5. Compute matrix-matrix multiplication: X^T * X
XTX = X.T @ X
print("\nXTX:\n", XTX)

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)
print("\nInverse of XTX:\n", XTX_inv)

# 7. Create target array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Compute weight vector w = (X^T * X)^(-1) * X^T * y
w = XTX_inv @ X.T @ y
print("\nWeight vector w:", w)

# 9. Calculate the sum of all elements in w
sum_of_weights = np.sum(w)
print(f"\nSum of all elements in w: {sum_of_weights:.4f}")
```
